```python
# prefix and postfix increment behavior
class Counter:
    def __init__(self, value):
        self.value = value

    def prefix_increment(self):
        self.value += 1
        return self.value

    def postfix_increment(self):
        temp = self.value
        self.value += 1
        return temp

c = Counter(5)
print(c.prefix_increment())   # Output: 6
print(c.postfix_increment())  # Output: 6 (but value becomes 7)
print(c.value)                # Final value: 7
```
